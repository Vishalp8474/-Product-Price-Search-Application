```python
from ddgs import DDGS                                                                  # Imports the DDGS class from the ddgs package.
from collections import defaultdict                                                    # Imports defaultdict from Python collections
import urllib.request                                                                  # Imports tools to open URLs and download webpage content.
from bs4 import BeautifulSoup                                                          # Imports BeautifulSoup, used for parsing HTML webpages                                                       


movie=input("Enter movie name: ")                                                       # Asks the user to enter a movie name.
t=[]                                                                                    # Creates an empty list named t
with DDGS() as ddgs:                                                                    # Creates a DuckDuckGo search object
    results = ddgs.text(movie+ "user review", max_results=5)                              # Searches DuckDuckGo for:

    for r in results:                                                                    #  Loops through each search result one by one. ,  Each r is usually a dictionary like:
        print(r['href'])                                                                #  Print URL
        s=r['href']                                                                     #  Saves the URL into variable s.
        t.append(s)                                                                     #  Adds the URL into list t. 
        print(s)
```

```python

import re                                                                                # re = Regular Expressions module.
h=[]                                                                                     # This list stores all extracted sentences.
for i in t:                                                                              # Loop visits each URL one by one.
    try:                                                                                 # Try Block
        response = urllib.request.urlopen(i)                                             # Connects to webpage URL.
        html = response.read()                                                           # Downloads webpage HTML.
        soup = BeautifulSoup(html)                                                       # BeautifulSoup converts raw HTML into a searchable structure.
        
        text = soup.get_text()                                                           # Removes HTML tags and keeps plain text.

        sentences = re.split(r' *[\.\?!][\'"\)\]]* *', text)                             # Split Text into Sentences
        for s in sentences:                                                              # Visits each sentence one by one.
            h.append(s)                                                                  # Adds sentence to list h

    except Exception as e:                                                               # Exception Handling
        print(e)
```

```python
h
```

```python
import pandas as pd
df=pd.DataFrame(h, columns=['Review'])
df
```

```python
def cleanTxt(text):
    text = re.sub(r'@[A-Za-z0-9]+', '', text)                                                    # Remove mentions
    text = re.sub(r'#', '', text)                                                                # Remove hashtag symbol
    text = re.sub(r'RT[\s]+', '', text)                                                          # Remove RT
    text = re.sub(r'https?:\/\/S+', '', text)                                                    # Remove URLs
    text = re.sub(r'0', '', text)                                                                # # Remove numbers
    text = re.sub(r'[0-9]', '', text)
    text = re.sub(r'%', '', text)                                                                # Remove %
    text = re.sub(r'\n', '', text)                                                               # Remove newlines
    return text
    
```

```python
df['Review'] = df['Review'].apply(cleanTxt)
df
```

```python
from afinn import Afinn
positive = []
negative = []
neutral = []

afinn = Afinn()

for review in df['Review']:
    if afinn.score(review) > 0:
        positive.append(review)
    elif afinn.score(review) <0:
        negative.append(review)
    elif afinn.score(review)==0:
        neutral.append(review)


print("Postive Reviews :", positive)
print(r'\t\n')
print("Negative Reviews :", negative)
print(r'\t\n')
print("neutral Reviews :", neutral)
print(r'\t\n')
total_reviews = len(df['Review'])
print("Total number of reviews :" ,total_reviews )
print("number of positive reviews :",  len(positive))
print("number of negative reviews :",  len(negative ))
print("number of neutural reviews :",  len(neutral))
```
